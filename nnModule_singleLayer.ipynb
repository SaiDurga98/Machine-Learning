```python
# Create model class
import torch
import torch.nn as nn

class Model(nn.Module):

  def __init__(self, num_features):
    # invokes the constructor pf parent class i.e.. nn module
    super().__init__()
    self.linear = nn.Linear(num_features, 1) # single neuron
    self.sigmoid = nn.Sigmoid() # Activation function

# forward pass
  def forward(self, features):
    output = self.linear(features) # z = wx + b
    output = self.sigmoid(output) # y_pred = sigmoid(z)
    return output
```

```python
# Create dataset : # 10 rows, 5 columns (features -> 5 )
features = torch.rand(10,5)
# Create Model
model = Model(features.shape[1])
# call foward pass
#model.forward(features)
model(features)
# Output will be y_pred for 10 rows
```
